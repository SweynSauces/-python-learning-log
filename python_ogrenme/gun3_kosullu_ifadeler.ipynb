If / Elif / Else 

```python
#if (eğer) - koşul doğruysa çalışacak kod bloğunu başlatır
#else (değilse) - yukarıdaki hiçbir koşul sağlanmadığında çalışacak son kaçış noktasıdır
#elif (değilse eğer) - iki koşul da yanlışsa alternatif bir koşul belirlemek için kullanılır
```

```python
#and (ve) - iki koşulun da doğru olması durumunda çalışacak kod bloğunu başlatır
#or (veya) - iki koşuldan en az birinin doğru olması durumunda
#not (değil) - koşulun tersini alır, koşul doğruysa yanlış, yanlışsa doğru olur
```

```python
#== (eşittir) - iki değerin eşit olup olmadığını kontrol eder
#!= (eşit değildir) - iki değerin eşit olmadığını kontrol eder
#> (büyüktür) - sol tarafın sağ taraftan büyük olup olmadığını kontrol eder
#< (küçüktür) - sol tarafın sağ taraftan küçük olup olmadığını kontrol eder
#>= (büyük veya eşittir) - sol tarafın sağ taraftan büyük veya eşit olup olmadığını kontrol eder
#<= (küçük veya eşittir) - sol tarafın sağ taraftan küçük veya eşit olup olmadığını kontrol eder
```

```python
not_ortalamasi = 95
print("Öğrencinin Not Ortalaması:", not_ortalamasi)
```

```python
if not_ortalamasi >= 90 and not_ortalamasi <= 100:
    print("Notunuz: AA - Mükemmel")
elif not_ortalamasi >= 80 and not_ortalamasi < 90:
    print("Notunuz: BA - Çok İyi")
elif not_ortalamasi >= 70 and not_ortalamasi < 80:
    print("Notunuz: BB - İyi")
elif not_ortalamasi >= 60 and not_ortalamasi < 70:
    print("Notunuz: CB - Orta")
elif not_ortalamasi >= 50 and not_ortalamasi < 60:
    print("Notunuz: CC - Geçer")
elif not_ortalamasi >= 0 and not_ortalamasi < 50:
    print("Notunuz: FF - Kaldınız")
else:
    print("Geçersiz not değeri. Lütfen 0 ile 100 arasında bir değer giriniz.")
```
