#LİSTE METODLARI

```python
envanter = ["Fener", "Cüzdan", "Anahtar"]
print("Başlangıç:", envanter)
```

```python
envanter.append("Telefon")       #append() fonksiyonu ile listenin sonuna yeni bir eleman ekleyebiliriz.
print("append sonrası:", envanter)
```

```python
envanter.insert(1, "Harita")     #insert() fonksiyonu ile listenin istediğimiz bir indexine yeni bir eleman ekleyebiliriz.
print("insert sonrası:", envanter)
```

```python
envanter.remove("Anahtar")       #remove() fonksiyonu ile listeden istediğimiz bir elemanı silebiliriz.
print("remove sonrası:", envanter)
```

```python
son_kullanilan = envanter.pop()  #pop() fonksiyonu ile listenin son elemanını silip o elemanı geri döndürebiliriz.
print("pop sonrası:", envanter)
print("pop ile çıkarılan eleman:", son_kullanilan)
```

```python
sayilar = [112,5,33,21,56]
sayilar.sort()                    #sort() fonksiyonu ile listeyi küçükten büyüğe sıralayabiliriz.
print("sort sonrası:", sayilar)
```

```python
sayilar.sort(reverse=True)       #reverse=True parametresi ile listeyi büyükten küçüğe sıralayabiliriz.
print("reverse sort sonrası:", sayilar)
```

```python
envanter.reverse()                    #reverse() fonksiyonu ile listenin elemanlarının sırasını tersine çevirebiliriz.
print("reverse sonrası envanter:", envanter)
```

```python
skorlar = [100, 30, 100, 60, 70,200]
yuz_sayisi = skorlar.count(100)         #count() fonksiyonu ile listede kaç tane 100 olduğunu öğrenebiliriz.
print(f"Listede {yuz_sayisi} tane 100 var.")
```

```python
envanter.clear()                        #clear() fonksiyonu ile listenin tüm elemanlarını silebiliriz.
print("clear sonrası envanter:", envanter)
```
