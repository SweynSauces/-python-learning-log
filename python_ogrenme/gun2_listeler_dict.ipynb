#LİSTE METODLARI

```python
envanter = ["Fener", "Cüzdan", "Anahtar"]
print("Başlangıç:", envanter)
```

```python
envanter.append("Telefon")       #append() fonksiyonu ile listenin sonuna yeni bir eleman ekleyebiliriz.
print("append sonrası:", envanter)
```

```python
envanter.insert(1, "Harita")     #insert() fonksiyonu ile listenin istediğimiz bir indexine yeni bir eleman ekleyebiliriz.
print("insert sonrası:", envanter)
```

```python
envanter.remove("Anahtar")       #remove() fonksiyonu ile listeden istediğimiz bir elemanı silebiliriz.
print("remove sonrası:", envanter)
```

```python
son_kullanilan = envanter.pop()  #pop() fonksiyonu ile listenin son elemanını silip o elemanı geri döndürebiliriz.
print("pop sonrası:", envanter)
print("pop ile çıkarılan eleman:", son_kullanilan)
```

```python
envanter = ["Fener", "Harita", "Cüzdan"]    # in operatörü ile listenin içinde istediğimiz bir elemanın olup olmadığını kontrol edebiliriz.
print("Harita" in envanter)                 #True döndürür çünkü Harita listede var.
print("Telefon" in envanter)                #False döndürür çünkü Telefon listede yok.
```

```python
sayilar = [112,5,33,21,56]
sayilar.sort()                    #sort() fonksiyonu ile listeyi küçükten büyüğe sıralayabiliriz.
print("sort sonrası:", sayilar)
```

```python
sayilar.sort(reverse=True)       #reverse=True parametresi ile listeyi büyükten küçüğe sıralayabiliriz.
print("reverse sort sonrası:", sayilar)
```

```python
envanter.reverse()                    #reverse() fonksiyonu ile listenin elemanlarının sırasını tersine çevirebiliriz.
print("reverse sonrası envanter:", envanter)
```

```python
skorlar = [100, 30, 100, 60, 70,200]
yuz_sayisi = skorlar.count(100)         #count() fonksiyonu ile listede kaç tane 100 olduğunu öğrenebiliriz.
print(f"Listede {yuz_sayisi} tane 100 var.")
```

```python
envanter.clear()                        #clear() fonksiyonu ile listenin tüm elemanlarını silebiliriz.
print("clear sonrası envanter:", envanter)
```

Dictionary

```python
#Listelerdeki gibi değerler yan yana yazılmaz, anahtar:değer eşleştirmesi vardır. Anahtarlar unique olmalıdır.
```

```python
kisi_bilgileri = {
    "isim": "Sercan",
    "yas": 21,
    "bolum": "Yönetim Bilişim Sistemleri",
    "ilgi_alanlari": ["Yapay Zeka", "Oyun Geliştirme"] # Sözlük içinde liste bile tutabiliriz
}
```

```python
print("Kişinin İsmi:", kisi_bilgileri["isim"])
print("Kişinin Yaşı:", kisi_bilgileri["yas"])
```

```python
print("Sözlüğün Anahtarları:")
print(kisi_bilgileri.keys())                        #keys() fonksiyonu ile sözlüğün anahtarlarını öğrenebiliriz.
```

```python
print("\nSözlüğün Değerleri:")
print(kisi_bilgileri.values())                      #values() fonksiyonu ile sözlüğün değerlerini öğrenebiliriz.
```

```python
print("\nSözlüğün Tüm Çiftleri:")
print(kisi_bilgileri.items())                       #.items() -> Anahtar ve değerleri birlikte tuple'lar halinde getirir (Döngülerde çok işe yarar)
```

```python
meyve_kalori = {"elma": 52, "muz": 89, "portakal": 47}
print("Meyve Kalorileri:")
print(meyve_kalori)
```

```python
meyve_kalori["elma"] = 55                       #Sözlükteki bir anahtarın değerini değiştirebiliriz.
print("Güncellenmiş Meyve Kalorileri:")
print(meyve_kalori)
```

```python
meyve_kalori["karpuz"] = 300                    #Sözlüğe yeni bir anahtar-değer çifti eklemek bu kadar basittir. Yeni bir anahtar eklediğimizde sözlüğün boyutu da artar.
print("Yeni eklenen Meyve Kalorileri:")
print(meyve_kalori)
```
