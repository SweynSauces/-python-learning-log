```python
name = "Sercan"         #Metin (String) veri tipi
age = 21                #Tam sayı (Integer) veri tipi
height = 1.85           #Ondalıklı sayı (Float) veri tipi
isMale = True           #Boolean veri tipi
```

```python

def kisiozellik():                              # def: bir fonksiyon tanımlar; kodu tekrar tekrar kullanmak için
    name = "Sercan"
    age = 21
    height = 1.85
    isMale = True
    return name, age, height, isMale

name, age, height, isMale = kisiozellik()
print(name)
print(age)
print(height)
print(isMale)                                 
```

#ARİTMETİK İŞLEMLER

#STRING BİRLEŞTİRME

```python
#Artı (+) Operatörü ile Birleştirme

ad = "Sercan"
yas = 21
hobi = "Video oyunları oynamak"
mesaj = "Adım " + ad + ", yaşım " + str(yas) + " ve hobim " + hobi + "."         #pythonda + işlemi veri tiplerine göre farklı davranır. String ve int veri tipleri birleştirilemez. Bu yüzden int veri tipi olan yaş değişkeni str() fonksiyonu ile string veri tipine dönüştürülür. 
print(mesaj)                                                                     
```

```python
#F-String Kullanımı

ad = "Sercan"
yas = 21
hobi = "Video oyunları oynamak"
mesaj = f"Adım {ad}, yaşım {yas} ve hobim {hobi}."         #f-string kullanımı ile değişkenler direkt olarak string içerisine yerleştirilebilir.
print(mesaj)
```

```python
#String veri tipinde bulunan değişkenler üzerinde bazı işlemler yapılabilir. 
#Örneğin, bir stringin tüm harflerini büyük harfe çevirmek için upper() fonksiyonu kullanılabilir. 
#Benzer şekilde, lower() fonksiyonu ile tüm harfler küçük harfe çevrilebilir ve capitalize() fonksiyonu ile sadece ilk harf büyük yapılabilir.


len(name)
print(len(name))
#len() fonksiyonu ile bir stringin uzunluğu (karakter sayısı) bulunabilir.

print(name.upper())
print(name.lower())
print(name.capitalize())
print(name.strip())
```

```python
age = 22
isMale = False

boolean = age > 18 and isMale == True           #iki şartın da aynı anda sağlanması durumunda True döndürür. Aksi takdirde False döndürür.
print(boolean)

boolean = age > 18 or isMale == True            #iki şarttan birinin sağlanması durumunda True döndürür. Aksi takdirde False döndürür.
print(boolean)
```
