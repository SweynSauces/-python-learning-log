Temel Fonksiyon Tanımlama 

```python
def selamla():                    # def: fonksiyon tanımlamanın anahtar kelimesi
    print("Merhaba, Python dünyasına hoş geldin!")

selamla()                          # fonksiyonu çağırmak, çalıştırmak için parantezle yazılır
# Dikkat: sadece "selamla" yazarsan (parantezsiz) fonksiyonu çalıştırmaz, 
# sadece fonksiyonun kendisini (bir obje olarak) gösterir.
```

```python
def selamla(isim):                 # isim: parametre, fonksiyon çağrılırken dışarıdan verilecek
    print(f"Merhaba, {isim}!")

selamla("Sercan")                  # "Sercan" burada argüman (parametreye verilen gerçek değer)
selamla("Ayşe")                    # aynı fonksiyon, farklı veriyle -> tekrar kullanılabilirlik
```

```python
def kisi_bilgisi(isim, yas):        
    print(f"{isim}, {yas} yaşında.")   #birden fazla parametre de alabilir.

kisi_bilgisi("Sercan", 21)         # sıraya göre eşleşir: isim="Sercan", yas=21
```

```python
#return

#print() sadece ekrana yazar, sonucu kullanamazsın. return ise fonksiyonun sonucunu dışarı "verir", sen bu sonucu bir değişkene atayıp başka işlemlerde kullanabilirsin. 

def topla(a, b):
    sonuc = a + b
    return sonuc                   # fonksiyonun ürettiği değeri dışarı "geri döndürür"

x = topla(5, 3)                    # dönen değeri x değişkenine atıyoruz
print(x)                           # 8
print(topla(10, 20) * 2)           # dönen değer doğrudan başka bir işlemde de kullanılabilir: 60

# fonksiyon içinde print() ile göstermek, dışarıda o değeri KULLANMANI sağlamaz.
# return kullanmadan bir fonksiyonun ürettiği veriyi dışarı taşıyamazsın.
# return çalıştığı anda fonksiyon hemen durur, altındaki kod çalışmaz.
```

```python
def kontrol(sayi):
    if sayi > 0:
        return "pozitif"           # burada return çalışırsa fonksiyon burada biter
    return "pozitif değil"         # sadece yukarıdaki koşul sağlanmazsa buraya gelinir

print(kontrol(5))                  # pozitif
print(kontrol(-3))                 # pozitif değil
```

```python
def selamla(isim, mesaj="Hoş geldin"):    # mesaj parametresi varsayılan olarak "Hoş geldin"
    print(f"{isim}, {mesaj}!")

selamla("Sercan")                          # mesaj verilmedi -> varsayılan kullanılır: "Sercan, Hoş geldin!"
selamla("Sercan", "Nasılsın")              # mesaj verildi -> varsayılan ezilir: "Sercan, Nasılsın!"

# KURAL: varsayılan değerli parametreler her zaman EN SONDA olmalı
# def selamla(mesaj="Hoş geldin", isim):   # BU HATA VERİR, yanlış sıralama
```

```python
def kisi_bilgileri():
    isim = "Sercan"
    yas = 21
    bolum = "YBS"
    return isim, yas, bolum          # Python bunu otomatik olarak bir tuple yapar: ("Sercan", 21, "YBS")


isim, yas, bolum = kisi_bilgileri()
print(isim)                          # Sercan
print(yas)                           # 21
```

```python
def kisi_bilgisi(isim, yas, bolum):
    print(f"{isim}, {yas} yaşında, {bolum} okuyor.")

kisi_bilgisi(yas=21, isim="Sercan", bolum="YBS")   # sıra karışık ama isimler doğru eşleşiyor
```

```python
def toplam_hesapla(*sayilar):             # *args: kaç tane argüman gelirse gelsin hepsini bir tuple'da toplar
    return sum(sayilar)                   # sum(): bir koleksiyondaki tüm sayıları toplayan hazır fonksiyon

print(toplam_hesapla(1, 2, 3))            # 6
print(toplam_hesapla(1, 2, 3, 4, 5))      # 15 -> aynı fonksiyon, farklı sayıda argümanla çalışıyor
```

```python
def kare_al(sayi):
    """Bir sayının karesini hesaplar ve döndürür."""    # docstring: fonksiyonun hemen altına yazılır
    return sayi ** 2

print(kare_al(5))                       # 25
```

```python
# TEHLİKELİ kullanım:
def liste_ekle(eleman, liste=[]):     # varsayılan değer olarak boş liste TEHLİKELİDİR
    liste.append(eleman)
    return liste

print(liste_ekle("a"))                 # ['a']
print(liste_ekle("b"))                 # ['a', 'b'] -> BEKLENMEDİK! Yeni boş liste değil, eskisi kullanıldı

# DOĞRU kullanım:
def liste_ekle(eleman, liste=None):
    if liste is None:
        liste = []
    liste.append(eleman)
    return liste
```
