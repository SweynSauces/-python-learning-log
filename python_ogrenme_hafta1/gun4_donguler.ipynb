Döngüler - For / While

```python
#FOR DÖNGÜSÜ
#Bilinen bir koleksiyon üzerinde (liste, tuple, dict, string, range) her eleman için sırayla bir işlem yapmak istediğinde kullanılır. 
#"Şu listedeki her eleman için şunu yap" mantığı. Kaç kez döneceğini önceden biliyorsun (koleksiyonun uzunluğu kadar).
```

```python
meyveler = ["elma", "armut", "muz"]
for meyve in meyveler:
    print(meyve)
```

```python
rutbeler = ["demir", "bronz", "gümüş", "altın", "plat", "zümrüt", "elmas", "usta", "büyükusta", "challenger"]
for rutbe in rutbeler:
    print(rutbe)
```

```python
for sayi in range(1,10):            #son sayı -10- dahil değil. tembel bir sayı üretici örneği
    print(sayi)
```

```python
for sayi1 in range(0,12,2):        #range fonksiyonu (başlangıç, bitiş, adım) şeklinde de yazılabilir
    print(sayi1)
```

```python
#WHILE DÖNGÜSÜ
#Bir koşul doğru olduğu sürece tekrar eder — kaç kez döneceğini önceden bilmiyorsun, sadece "ne zaman durması gerektiğini" biliyorsun. 
#for'dan temel farkı: for "şu kadar kere dön", while "şu doğru olduğu sürece dön".
```

```python
stamina = 20
while stamina >0:
    print("Enerji: ", stamina)
    stamina -=1                     #Kritik nokta; eğer 'stamina -=1' koşulunu eklemeseydik program sonsuz dönecekti. 
```

```python
for sayi in range(1, 20):
    if sayi == 7:
        break              # 7'ye gelince döngü tamamen durur
    print(sayi)
```

```python
for sayi in range(1, 10):
    if sayi % 2 == 0:
        continue           # çift sayıları atla, döngü devam etsin
    print(sayi)            # sadece tek sayılar yazdırılır
```

```python
for i in range(1,6):
    for j in range (1,6):   #genellikle satır-sütun araması gerektiren yerlerde kullanılır
        print(i,j)
```

```python
envanter = ["Fener", "Harita", "Anahtar"]
for index, esya in enumerate(envanter):     #Bazen listelerde hem eleman hem de o elemanın indexi gerekebilir. Bu tip durumlarda sayaç tutmak gerekir ama enumare() bunu otomatikleştirir.
    print(index, "-", esya)
```

```python
isimler = ["Ali", "Ayşe", "Mert"]
yaslar = [25, 30, 22]
for isim, yas in zip(isimler, yaslar):          #İki listeyi de aynı anda paralel biçimde gezmek için kullanılır.
    print(isim, yas)
```
